```python
"""
Script para consultar y visualizar los reportes generados
Autor: [Tu Nombre]
Fecha: 2025-10-18
"""

import pandas as pd
import sqlite3
import json
import os
from pymongo import MongoClient

print("="*80)
print("CONSULTA DE REPORTES GENERADOS")
print("="*80)

# =============================================================================
# REPORTE 1: CSV - Top Vinos por Continente
# =============================================================================

print("\n📊 REPORTE 1: Top Vinos por Continente (CSV)")
print("-"*80)

if os.path.exists('reportes/reporte1_top_vinos_continente.csv'):
    reporte1 = pd.read_csv('reportes/reporte1_top_vinos_continente.csv')
    
    print(f"\n✓ Archivo encontrado: {len(reporte1)} registros")
    print("\nPrimeros 10 registros:")
    print(reporte1.head(10))
    
    print("\nDistribución por continente:")
    if 'continente' in reporte1.columns:
        print(reporte1['continente'].value_counts())
    
    # Análisis adicional
    print("\n📈 Estadísticas:")
    print(f"  - Puntuación máxima: {reporte1['puntuacion'].max()}")
    print(f"  - Puntuación mínima: {reporte1['puntuacion'].min()}")
    print(f"  - Puntuación promedio: {reporte1['puntuacion'].mean():.2f}")
    
    if 'precio' in reporte1.columns:
        precio_sin_nulos = reporte1['precio'].dropna()
        if len(precio_sin_nulos) > 0:
            print(f"  - Precio promedio: ${precio_sin_nulos.mean():.2f}")
            print(f"  - Precio máximo: ${precio_sin_nulos.max():.2f}")
else:
    print("✗ Archivo no encontrado. Ejecuta primero el script principal.")

# =============================================================================
# REPORTE 2: Excel - Estadísticas por País
# =============================================================================

print("\n\n📊 REPORTE 2: Estadísticas por País (Excel)")
print("-"*80)

if os.path.exists('reportes/reporte2_estadisticas_pais.xlsx'):
    reporte2 = pd.read_excel('reportes/reporte2_estadisticas_pais.xlsx')
    
    print(f"\n✓ Archivo encontrado: {len(reporte2)} países analizados")
    
    print("\n🏆 Top 10 países con mejor puntuación promedio:")
    print(reporte2.head(10).to_string(index=False))
    
    print("\n📊 Top 10 países con más reviews:")
    top_reviews = reporte2.nlargest(10, 'cantidad_reviews')
    print(top_reviews[['country', 'cantidad_reviews', 'puntuacion_promedio']].to_string(index=False))
    
    print("\n💰 Top 10 países con vinos más caros:")
    top_caros = reporte2.nlargest(10, 'precio_promedio')
    print(top_caros[['country', 'precio_promedio', 'puntuacion_promedio']].to_string(index=False))
    
    # Estadísticas generales
    print("\n📈 Estadísticas Generales:")
    print(f"  - Total de países: {len(reporte2)}")
    print(f"  - Total de reviews: {reporte2['cantidad_reviews'].sum()}")
    print(f"  - Precio promedio global: ${reporte2['precio_promedio'].mean():.2f}")
    print(f"  - Puntuación promedio global: {reporte2['puntuacion_promedio'].mean():.2f}")
else:
    print("✗ Archivo no encontrado. Ejecuta primero el script principal.")

# =============================================================================
# REPORTE 3: SQLite - Variedades de Uva
# =============================================================================

print("\n\n📊 REPORTE 3: Variedades de Uva Más Populares (SQLite)")
print("-"*80)

if os.path.exists('reportes/reporte3_variedades.db'):
    conn = sqlite3.connect('reportes/reporte3_variedades.db')
    
    # Consultar todas las variedades
    reporte3 = pd.read_sql_query("SELECT * FROM variedades_populares", conn)
    
    print(f"\n✓ Base de datos encontrada: {len(reporte3)} variedades")
    
    print("\n🍇 Top 10 variedades más populares:")
    print(reporte3.head(10).to_string(index=False))
    
    # Consultas SQL personalizadas
    print("\n🔍 Variedades con puntuación > 90:")
    query_alta_puntuacion = """
    SELECT variedad, cantidad, puntuacion_promedio, precio_promedio
    FROM variedades_populares
    WHERE puntuacion_promedio > 90
    ORDER BY puntuacion_promedio DESC
    """
    alta_puntuacion = pd.read_sql_query(query_alta_puntuacion, conn)
    print(alta_puntuacion.to_string(index=False))
    
    print("\n💎 Variedades más caras (precio > $50):")
    query_caras = """
    SELECT variedad, cantidad, precio_promedio, puntuacion_promedio
    FROM variedades_populares
    WHERE precio_promedio > 50
    ORDER BY precio_promedio DESC
    LIMIT 10
    """
    caras = pd.read_sql_query(query_caras, conn)
    print(caras.to_string(index=False))
    
    conn.close()
else:
    print("✗ Base de datos no encontrada. Ejecuta primero el script principal.")

# =============================================================================
# REPORTE 4: JSON/MongoDB - Mejores Bodegas
# =============================================================================

print("\n\n📊 REPORTE 4: Mejores Bodegas (JSON/MongoDB)")
print("-"*80)

# Intentar leer de MongoDB primero
mongodb_disponible = False
try:
    client = MongoClient('mongodb://localhost:27017/', serverSelectionTimeoutMS=2000)
    client.server_info()
    mongodb_disponible = True
    
    db = client['wine_database']
    collection = db['mejores_bodegas']
    
    documentos = list(collection.find({}, {'_id': 0}))
    
    if documentos:
        reporte4 = pd.DataFrame(documentos)
        print(f"\n✓ MongoDB conectado: {len(reporte4)} bodegas encontradas")
        
        print("\n🏭 Top 10 bodegas con más vinos excelentes:")
        print(reporte4.head(10).to_string(index=False))
        
        print("\n📊 Bodegas por país:")
        if 'pais_principal' in reporte4.columns:
            print(reporte4['pais_principal'].value_counts().head(10))
    else:
        print("⚠ MongoDB conectado pero colección vacía")
        mongodb_disponible = False
    
    client.close()
    
except Exception as e:
    print(f"⚠ MongoDB no disponible: {str(e)}")
    mongodb_disponible = False

# Si MongoDB no está disponible, leer desde JSON
if not mongodb_disponible:
    print("\n📁 Leyendo desde archivo JSON...")
    
    if os.path.exists('reportes/reporte4_mejores_bodegas.json'):
        with open('reportes/reporte4_mejores_bodegas.json', 'r', encoding='utf-8') as f:
            datos_json = json.load(f)
        
        reporte4 = pd.DataFrame(datos_json)
        
        print(f"\n✓ Archivo JSON encontrado: {len(reporte4)} bodegas")
        
        print("\n🏭 Top 10 bodegas con más vinos excelentes:")
        print(reporte4.head(10).to_string(index=False))
        
        print("\n📊 Bodegas por país:")
        if 'pais_principal' in reporte4.columns:
            print(reporte4['pais_principal'].value_counts().head(10))
        
        print("\n💰 Bodegas con vinos más caros:")
        top_caras_bodegas = reporte4.nlargest(10, 'precio_promedio')
        print(top_caras_bodegas[['bodega', 'precio_promedio', 'puntuacion_promedio']].to_string(index=False))
    else:
        print("✗ Archivo JSON no encontrado. Ejecuta primero el script principal.")

# =============================================================================
# COMPARACIÓN ENTRE REPORTES
# =============================================================================

print("\n\n📊 ANÁLISIS COMPARATIVO")
print("-"*80)

try:
    # Verificar qué reportes existen
    reportes_existentes = []
    
    if os.path.exists('reportes/reporte1_top_vinos_continente.csv'):
        reportes_existentes.append("Reporte 1 (CSV)")
    if os.path.exists('reportes/reporte2_estadisticas_pais.xlsx'):
        reportes_existentes.append("Reporte 2 (Excel)")
    if os.path.exists('reportes/reporte3_variedades.db'):
        reportes_existentes.append("Reporte 3 (SQLite)")
    if os.path.exists('reportes/reporte4_mejores_bodegas.json'):
        reportes_existentes.append("Reporte 4 (JSON)")
    
    print("\n✓ Reportes disponibles:")
    for reporte in reportes_existentes:
        print(f"  - {reporte}")
    
    print(f"\n📈 Total de reportes generados: {len(reportes_existentes)}/4")
    
    if len(reportes_existentes) == 4:
        print("\n✅ ¡Todos los reportes están disponibles y listos para consultar!")
    else:
        print("\n⚠ Algunos reportes faltan. Ejecuta el script principal para generarlos.")
    
except Exception as e:
    print(f"\n✗ Error al verificar reportes: {str(e)}")

# =============================================================================
# INSTRUCCIONES ADICIONALES
# =============================================================================

print("\n\n" + "="*80)
print("INSTRUCCIONES PARA ANÁLISIS ADICIONAL")
print("="*80)

print("""
📚 Puedes realizar análisis adicionales con los reportes:

1. FILTRAR DATOS:
   df = pd.read_csv('reportes/reporte1_top_vinos_continente.csv')
   vinos_francia = df[df['country'] == 'France']

2. CREAR GRÁFICOS:
   import matplotlib.pyplot as plt
   df['puntuacion'].hist(bins=20)
   plt.show()

3. CONSULTAS SQL PERSONALIZADAS:
   conn = sqlite3.connect('reportes/reporte3_variedades.db')
   query = "SELECT * FROM variedades_populares WHERE cantidad > 1000"
   resultado = pd.read_sql_query(query, conn)

4. EXPORTAR A OTROS FORMATOS:
   df.to_json('mi_analisis.json')
   df.to_html('mi_analisis.html')

5. COMBINAR REPORTES:
   reporte1 = pd.read_csv('reportes/reporte1_top_vinos_continente.csv')
   reporte2 = pd.read_excel('reportes/reporte2_estadisticas_pais.xlsx')
   combinado = pd.merge(reporte1, reporte2, on='country')
""")

print("\n" + "="*80)
print("CONSULTA COMPLETADA")
print("="*80)
```

```python
"""
Script para enviar reportes por correo electrónico
Autor: [Tu Nombre]
Fecha: 2025-10-18

IMPORTANTE: Antes de ejecutar este script:
1. Habilita la verificación en 2 pasos en tu cuenta de Google
2. Genera una contraseña de aplicación en: https://myaccount.google.com/apppasswords
3. Completa las variables de configuración abajo
"""

import smtplib
from email.mime.multipart import MIMEMultipart
from email.mime.text import MIMEText
from email.mime.base import MIMEBase
from email import encoders
import os
from datetime import datetime

# =============================================================================
# CONFIGURACIÓN - MODIFICA ESTOS VALORES CON TUS DATOS
# =============================================================================

# Configuración del remitente (tu cuenta de Gmail)
REMITENTE = "tu_email@gmail.com"
PASSWORD_APP = "xxxx xxxx xxxx xxxx"  # Contraseña de aplicación de 16 caracteres

# Configuración del destinatario
DESTINATARIO = "destinatario@email.com"

# Configuración del servidor SMTP
SMTP_SERVER = "smtp.gmail.com"
SMTP_PORT = 587

# =============================================================================
# FUNCIÓN PARA ENVIAR CORREO
# =============================================================================

def enviar_correo_con_adjunto(archivo_adjunto, asunto, destinatario, remitente, password):
    """
    Envía un correo electrónico con archivo adjunto
    
    Parámetros:
    -----------
    archivo_adjunto : str
        Ruta del archivo a adjuntar
    asunto : str
        Asunto del correo
    destinatario : str
        Email del destinatario
    remitente : str
        Email del remitente (tu cuenta)
    password : str
        Contraseña de aplicación de Gmail
    
    Retorna:
    --------
    bool : True si se envió exitosamente, False en caso contrario
    """
    
    try:
        # Verificar que el archivo existe
        if not os.path.exists(archivo_adjunto):
            print(f"✗ Error: El archivo {archivo_adjunto} no existe")
            return False
        
        # Crear el mensaje
        mensaje = MIMEMultipart()
        mensaje['From'] = remitente
        mensaje['To'] = destinatario
        mensaje['Subject'] = asunto
        
        # Crear el cuerpo del correo en HTML
        cuerpo_html = f"""
        <html>
        <head>
            <style>
                body {{
                    font-family: Arial, sans-serif;
                    line-height: 1.6;
                    color: #333;
                }}
                .header {{
                    background-color: #722f37;
                    color: white;
                    padding: 20px;
                    text-align: center;
                }}
                .content {{
                    padding: 20px;
                    background-color: #f4f4f4;
                }}
                .info-box {{
                    background-color: white;
                    border-left: 4px solid #722f37;
                    padding: 15px;
                    margin: 10px 0;
                }}
                .footer {{
                    background-color: #333;
                    color: white;
                    padding: 10px;
                    text-align: center;
                    font-size: 12px;
                }}
                table {{
                    border-collapse: collapse;
                    width: 100%;
                    margin: 15px 0;
                }}
                th, td {{
                    border: 1px solid #ddd;
                    padding: 8px;
                    text-align: left;
                }}
                th {{
                    background-color: #722f37;
                    color: white;
                }}
            </style>
        </head>
        <body>
            <div class="header">
                <h1>🍷 Reporte de Análisis de Vinos</h1>
                <p>Wine Magazine - 130K Reviews</p>
            </div>
            
            <div class="content">
                <p><strong>Estimado/a,</strong></p>
                
                <p>Adjunto encontrará el reporte de análisis de datos de vinos del Wine Magazine, 
                generado automáticamente por nuestro sistema de análisis.</p>
                
                <div class="info-box">
                    <h3>📊 Información del Reporte</h3>
                    <table>
                        <tr>
                            <th>Archivo</th>
                            <td>{os.path.basename(archivo_adjunto)}</td>
                        </tr>
                        <tr>
                            <th>Tamaño</th>
                            <td>{os.path.getsize(archivo_adjunto) / 1024:.2f} KB</td>
                        </tr>
                        <tr>
                            <th>Fecha de generación</th>
                            <td>{datetime.now().strftime('%Y-%m-%d %H:%M:%S')}</td>
                        </tr>
                    </table>
                </div>
                
                <div class="info-box">
                    <h3>📝 Contenido del Reporte</h3>
                    <p>Este reporte contiene:</p>
                    <ul>
                        <li>Estadísticas detalladas por país</li>
                        <li>Precio promedio de vinos por región</li>
                        <li>Puntuación promedio por país</li>
                        <li>Cantidad total de reviews analizadas</li>
                    </ul>
                </div>
                
                <div class="info-box">
                    <h3>💡 Recomendaciones</h3>
                    <p>Para visualizar el reporte:</p>
                    <ol>
                        <li>Descargue el archivo adjunto</li>
                        <li>Abra con Microsoft Excel, LibreOffice Calc o Google Sheets</li>
                        <li>Explore las diferentes estadísticas presentadas</li>
                    </ol>
                </div>
                
                <p>Si tiene alguna pregunta o necesita información adicional, no dude en contactarnos.</p>
                
                <p><strong>Saludos cordiales,</strong><br>
                <em>Sistema Automatizado de Análisis de Vinos</em><br>
                <em>Wine Data Analytics Team</em></p>
            </div>
            
            <div class="footer">
                <p>Este es un correo automático generado por el sistema de análisis de datos.</p>
                <p>© 2025 Wine Analytics | Todos los derechos reservados</p>
            </div>
        </body>
        </html>
        """
        
        # Adjuntar el cuerpo HTML al mensaje
        mensaje.attach(MIMEText(cuerpo_html, 'html'))
        
        # Adjuntar el archivo
        nombre_archivo = os.path.basename(archivo_adjunto)
        
        with open(archivo_adjunto, 'rb') as archivo:
            parte = MIMEBase('application', 'octet-stream')
            parte.set_payload(archivo.read())
        
        # Codificar el archivo en base64
        encoders.encode_base64(parte)
        
        # Agregar header al archivo adjunto
        parte.add_header(
            'Content-Disposition',
            f'attachment; filename= {nombre_archivo}'
        )
        
        # Adjuntar el archivo al mensaje
        mensaje.attach(parte)
        
        # Crear conexión SMTP segura
        print(f"\n📧 Conectando al servidor {SMTP_SERVER}:{SMTP_PORT}...")
        servidor = smtplib.SMTP(SMTP_SERVER, SMTP_PORT)
        servidor.starttls()  # Habilitar seguridad TLS
        
        print(f"🔐 Autenticando como {remitente}...")
        servidor.login(remitente, password)
        
        print(f"📤 Enviando correo a {destinatario}...")
        texto = mensaje.as_string()
        servidor.sendmail(remitente, destinatario, texto)
        
        print("✅ ¡Correo enviado exitosamente!")
        
        # Cerrar conexión
        servidor.quit()
        
        return True
        
    except smtplib.SMTPAuthenticationError:
        print("\n✗ Error de autenticación:")
        print("  - Verifica que el email sea correcto")
        print("  - Asegúrate de usar una contraseña de aplicación (no tu contraseña normal)")
        print("  - Verifica que la verificación en 2 pasos esté habilitada")
        return False
        
    except smtplib.SMTPException as e:
        print(f"\n✗ Error SMTP: {str(e)}")
        return False
        
    except Exception as e:
        print(f"\n✗ Error inesperado: {str(e)}")
        return False

# =============================================================================
# MENÚ INTERACTIVO PARA SELECCIONAR REPORTE
# =============================================================================

def menu_principal():
    """
    Muestra un menú para seleccionar qué reporte enviar
    """
    
    print("\n" + "="*80)
    print("ENVÍO DE REPORTES POR CORREO ELECTRÓNICO")
    print("="*80)
    
    # Verificar configuración
    if REMITENTE == "tu_email@gmail.com" or PASSWORD_APP == "xxxx xxxx xxxx xxxx":
        print("\n⚠️  ADVERTENCIA: Configuración incompleta")
        print("="*80)
        print("\nDebes configurar las siguientes variables en el script:")
        print("  - REMITENTE: Tu dirección de Gmail")
        print("  - PASSWORD_APP: Tu contraseña de aplicación de Gmail")
        print("  - DESTINATARIO: Email del destinatario")
        print("\n📚 Pasos para obtener contraseña de aplicación:")
        print("  1. Ve a: https://myaccount.google.com/apppasswords")
        print("  2. Selecciona 'Correo' y 'Otro dispositivo'")
        print("  3. Copia la contraseña generada (16 caracteres con espacios)")
        print("\n" + "="*80)
        return
    
    # Buscar reportes disponibles
    reportes_disponibles = []
    
    reportes = [
        ('reportes/reporte1_top_vinos_continente.csv', 'Reporte 1: Top Vinos por Continente (CSV)'),
        ('reportes/reporte2_estadisticas_pais.xlsx', 'Reporte 2: Estadísticas por País (Excel)'),
        ('reportes/reporte3_variedades.db', 'Reporte 3: Variedades Populares (SQLite)'),
        ('reportes/reporte4_mejores_bodegas.json', 'Reporte 4: Mejores Bodegas (JSON)')
    ]
    
    for ruta, descripcion in reportes:
        if os.path.exists(ruta):
            reportes_disponibles.append((ruta, descripcion))
    
    if not reportes_disponibles:
        print("\n✗ No se encontraron reportes.")
        print("  Ejecuta primero el script principal 'analisis_vinos.py'")
        return
    
    print(f"\n✓ Configuración:")
    print(f"  Remitente: {REMITENTE}")
    print(f"  Destinatario: {DESTINATARIO}")
    
    print(f"\n📁 Reportes disponibles: {len(reportes_disponibles)}")
    
    for i, (ruta, descripcion) in enumerate(reportes_disponibles, 1):
        tamaño = os.path.getsize(ruta) / 1024
        print(f"  {i}. {descripcion}")
        print(f"     Archivo: {os.path.basename(ruta)} ({tamaño:.2f} KB)")
    
    print(f"  {len(reportes_disponibles) + 1}. Enviar todos los reportes")
    print("  0. Salir")
    
    try:
        opcion = int(input("\n📌 Selecciona una opción: "))
        
        if opcion == 0:
            print("\n👋 Saliendo...")
            return
        
        elif 1 <= opcion <= len(reportes_disponibles):
            # Enviar un reporte específico
            archivo, descripcion = reportes_disponibles[opcion - 1]
            asunto = f"Reporte de Análisis de Vinos - {descripcion}"
            
            print(f"\n📤 Enviando: {descripcion}")
            
            if enviar_correo_con_adjunto(archivo, asunto, DESTINATARIO, REMITENTE, PASSWORD_APP):
                print("\n" + "="*80)
                print("✅ CORREO ENVIADO EXITOSAMENTE")
                print("="*80)
                print(f"\n📧 Revisa la bandeja de entrada de: {DESTINATARIO}")
                print(f"📎 Archivo adjunto: {os.path.basename(archivo)}")
                print("\n💡 Toma un pantallazo del correo recibido para tu entrega.")
            
        elif opcion == len(reportes_disponibles) + 1:
            # Enviar todos los reportes
            print("\n📤 Enviando todos los reportes...")
            exitosos = 0
            
            for archivo, descripcion in reportes_disponibles:
                asunto = f"Reporte de Análisis de Vinos - {descripcion}"
                print(f"\n📧 Enviando: {descripcion}")
                
                if enviar_correo_con_adjunto(archivo, asunto, DESTINATARIO, REMITENTE, PASSWORD_APP):
                    exitosos += 1
                    print(f"  ✓ Enviado: {os.path.basename(archivo)}")
                else:
                    print(f"  ✗ Error al enviar: {os.path.basename(archivo)}")
            
            print("\n" + "="*80)
            print(f"✅ PROCESO COMPLETADO: {exitosos}/{len(reportes_disponibles)} reportes enviados")
            print("="*80)
        
        else:
            print("\n✗ Opción inválida")
    
    except ValueError:
        print("\n✗ Por favor, ingresa un número válido")
    except KeyboardInterrupt:
        print("\n\n👋 Proceso cancelado por el usuario")

# =============================================================================
# EJECUCIÓN DEL SCRIPT
# =============================================================================

if __name__ == "__main__":
    menu_principal()
```

```python
"""
Script para enviar reportes por correo electronico
Sin caracteres especiales para evitar errores de sintaxis
"""

import smtplib
from email.mime.multipart import MIMEMultipart
from email.mime.text import MIMEText
from email.mime.base import MIMEBase
from email import encoders
import os
from datetime import datetime

# =============================================================================
# CONFIGURACION - MODIFICA ESTOS VALORES CON TUS DATOS
# =============================================================================

# Tu cuenta de Gmail
REMITENTE = "alcalaladrian634@gmail.com"

# Contrasena de aplicacion de Gmail (16 caracteres)
PASSWORD_APP = "123456"

# Email del destinatario
DESTINATARIO = "adrianalcala063@gmail.com"

# Servidor SMTP de Gmail
SMTP_SERVER = "smtp.gmail.com"
SMTP_PORT = 587

# =============================================================================
# FUNCION PARA ENVIAR CORREO
# =============================================================================

def enviar_correo_con_adjunto(archivo_adjunto, asunto, destinatario, remitente, password):
    """
    Envia un correo electronico con archivo adjunto
    """
    
    try:
        # Verificar que el archivo existe
        if not os.path.exists(archivo_adjunto):
            print("Error: El archivo {} no existe".format(archivo_adjunto))
            return False
        
        # Crear el mensaje
        mensaje = MIMEMultipart()
        mensaje['From'] = remitente
        mensaje['To'] = destinatario
        mensaje['Subject'] = asunto
        
        # Crear el cuerpo del correo en HTML
        cuerpo_html = """
        <html>
        <head>
            <style>
                body {{
                    font-family: Arial, sans-serif;
                    line-height: 1.6;
                    color: #333;
                }}
                .header {{
                    background-color: #722f37;
                    color: white;
                    padding: 20px;
                    text-align: center;
                }}
                .content {{
                    padding: 20px;
                    background-color: #f4f4f4;
                }}
                .info-box {{
                    background-color: white;
                    border-left: 4px solid #722f37;
                    padding: 15px;
                    margin: 10px 0;
                }}
                .footer {{
                    background-color: #333;
                    color: white;
                    padding: 10px;
                    text-align: center;
                    font-size: 12px;
                }}
                table {{
                    border-collapse: collapse;
                    width: 100%;
                    margin: 15px 0;
                }}
                th, td {{
                    border: 1px solid #ddd;
                    padding: 8px;
                    text-align: left;
                }}
                th {{
                    background-color: #722f37;
                    color: white;
                }}
            </style>
        </head>
        <body>
            <div class="header">
                <h1>Reporte de Analisis de Vinos</h1>
                <p>Wine Magazine - 130K Reviews</p>
            </div>
            
            <div class="content">
                <p><strong>Estimado/a,</strong></p>
                
                <p>Adjunto encontrara el reporte de analisis de datos de vinos del Wine Magazine.</p>
                
                <div class="info-box">
                    <h3>Informacion del Reporte</h3>
                    <table>
                        <tr>
                            <th>Archivo</th>
                            <td>{archivo}</td>
                        </tr>
                        <tr>
                            <th>Tamano</th>
                            <td>{tamano} KB</td>
                        </tr>
                        <tr>
                            <th>Fecha de generacion</th>
                            <td>{fecha}</td>
                        </tr>
                    </table>
                </div>
                
                <div class="info-box">
                    <h3>Contenido del Reporte</h3>
                    <p>Este reporte contiene:</p>
                    <ul>
                        <li>Estadisticas detalladas por pais</li>
                        <li>Precio promedio de vinos por region</li>
                        <li>Puntuacion promedio por pais</li>
                        <li>Cantidad total de reviews analizadas</li>
                    </ul>
                </div>
                
                <p>Saludos cordiales,</p>
                <p><em>Sistema de Analisis de Vinos</em></p>
            </div>
            
            <div class="footer">
                <p>Correo automatico generado por el sistema de analisis de datos.</p>
                <p>2025 Wine Analytics</p>
            </div>
        </body>
        </html>
        """.format(
            archivo=os.path.basename(archivo_adjunto),
            tamano=os.path.getsize(archivo_adjunto) / 1024,
            fecha=datetime.now().strftime('%Y-%m-%d %H:%M:%S')
        )
        
        # Adjuntar el cuerpo HTML al mensaje
        mensaje.attach(MIMEText(cuerpo_html, 'html'))
        
        # Adjuntar el archivo
        nombre_archivo = os.path.basename(archivo_adjunto)
        
        with open(archivo_adjunto, 'rb') as archivo:
            parte = MIMEBase('application', 'octet-stream')
            parte.set_payload(archivo.read())
        
        # Codificar el archivo en base64
        encoders.encode_base64(parte)
        
        # Agregar header al archivo adjunto
        parte.add_header(
            'Content-Disposition',
            'attachment; filename= {}'.format(nombre_archivo)
        )
        
        # Adjuntar el archivo al mensaje
        mensaje.attach(parte)
        
        # Crear conexion SMTP segura
        print("\nConectando al servidor {}:{}...".format(SMTP_SERVER, SMTP_PORT))
        servidor = smtplib.SMTP(SMTP_SERVER, SMTP_PORT)
        servidor.starttls()
        
        print("Autenticando como {}...".format(remitente))
        servidor.login(remitente, password)
        
        print("Enviando correo a {}...".format(destinatario))
        texto = mensaje.as_string()
        servidor.sendmail(remitente, destinatario, texto)
        
        print("CORREO ENVIADO EXITOSAMENTE!")
        
        # Cerrar conexion
        servidor.quit()
        
        return True
        
    except smtplib.SMTPAuthenticationError:
        print("\nError de autenticacion:")
        print("  - Verifica que el email sea correcto")
        print("  - Asegurate de usar una contrasena de aplicacion")
        print("  - Verifica que la verificacion en 2 pasos este habilitada")
        return False
        
    except smtplib.SMTPException as e:
        print("\nError SMTP: {}".format(str(e)))
        return False
        
    except Exception as e:
        print("\nError inesperado: {}".format(str(e)))
        return False

# =============================================================================
# MENU INTERACTIVO
# =============================================================================

def menu_principal():
    """
    Menu para seleccionar que reporte enviar
    """
    
    print("\n" + "="*80)
    print("ENVIO DE REPORTES POR CORREO ELECTRONICO")
    print("="*80)
    
    # Verificar configuracion
    if REMITENTE == "tu_email@gmail.com" or PASSWORD_APP == "xxxx xxxx xxxx xxxx":
        print("\nADVERTENCIA: Configuracion incompleta")
        print("="*80)
        print("\nDebes configurar las siguientes variables en el script:")
        print("  - REMITENTE: Tu direccion de Gmail")
        print("  - PASSWORD_APP: Tu contrasena de aplicacion de Gmail")
        print("  - DESTINATARIO: Email del destinatario")
        print("\nPasos para obtener contrasena de aplicacion:")
        print("  1. Ve a: https://myaccount.google.com/apppasswords")
        print("  2. Selecciona 'Correo' y 'Otro dispositivo'")
        print("  3. Copia la contrasena generada (16 caracteres con espacios)")
        print("\n" + "="*80)
        return
    
    # Buscar reportes disponibles
    reportes_disponibles = []
    
    reportes = [
        ('reportes/reporte1_top_vinos_continente.csv', 'Reporte 1: Top Vinos por Continente (CSV)'),
        ('reportes/reporte2_estadisticas_pais.xlsx', 'Reporte 2: Estadisticas por Pais (Excel)'),
        ('reportes/reporte3_variedades.db', 'Reporte 3: Variedades Populares (SQLite)'),
        ('reportes/reporte4_mejores_bodegas.json', 'Reporte 4: Mejores Bodegas (JSON)')
    ]
    
    for ruta, descripcion in reportes:
        if os.path.exists(ruta):
            reportes_disponibles.append((ruta, descripcion))
    
    if not reportes_disponibles:
        print("\nNo se encontraron reportes.")
        print("Ejecuta primero el script principal 'analisis_vinos.py'")
        return
    
    print("\nConfiguracion:")
    print("  Remitente: {}".format(REMITENTE))
    print("  Destinatario: {}".format(DESTINATARIO))
    
    print("\nReportes disponibles: {}".format(len(reportes_disponibles)))
    
    for i, (ruta, descripcion) in enumerate(reportes_disponibles, 1):
        tamano = os.path.getsize(ruta) / 1024
        print("  {}. {}".format(i, descripcion))
        print("     Archivo: {} ({:.2f} KB)".format(os.path.basename(ruta), tamano))
    
    print("  {}. Enviar todos los reportes".format(len(reportes_disponibles) + 1))
    print("  0. Salir")
    
    try:
        opcion = int(input("\nSelecciona una opcion: "))
        
        if opcion == 0:
            print("\nSaliendo...")
            return
        
        elif 1 <= opcion <= len(reportes_disponibles):
            # Enviar un reporte especifico
            archivo, descripcion = reportes_disponibles[opcion - 1]
            asunto = "Reporte de Analisis de Vinos - {}".format(descripcion)
            
            print("\nEnviando: {}".format(descripcion))
            
            if enviar_correo_con_adjunto(archivo, asunto, DESTINATARIO, REMITENTE, PASSWORD_APP):
                print("\n" + "="*80)
                print("CORREO ENVIADO EXITOSAMENTE")
                print("="*80)
                print("\nRevisa la bandeja de entrada de: {}".format(DESTINATARIO))
                print("Archivo adjunto: {}".format(os.path.basename(archivo)))
                print("\nToma un pantallazo del correo recibido para tu entrega.")
            
        elif opcion == len(reportes_disponibles) + 1:
            # Enviar todos los reportes
            print("\nEnviando todos los reportes...")
            exitosos = 0
            
            for archivo, descripcion in reportes_disponibles:
                asunto = "Reporte de Analisis de Vinos - {}".format(descripcion)
                print("\nEnviando: {}".format(descripcion))
                
                if enviar_correo_con_adjunto(archivo, asunto, DESTINATARIO, REMITENTE, PASSWORD_APP):
                    exitosos += 1
                    print("  Enviado: {}".format(os.path.basename(archivo)))
                else:
                    print("  Error al enviar: {}".format(os.path.basename(archivo)))
            
            print("\n" + "="*80)
            print("PROCESO COMPLETADO: {}/{} reportes enviados".format(exitosos, len(reportes_disponibles)))
            print("="*80)
        
        else:
            print("\nOpcion invalida")
    
    except ValueError:
        print("\nPor favor, ingresa un numero valido")
    except KeyboardInterrupt:
        print("\n\nProceso cancelado por el usuario")

# =============================================================================
# EJECUCION DEL SCRIPT
# =============================================================================

if __name__ == "__main__":
    menu_principal()
```
